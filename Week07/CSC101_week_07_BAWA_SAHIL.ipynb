# My Data Story: Books I Have Read
## Week 8 - Files and Friends

For this assignment I made a text file with books I have read, as well as one book I am currently reading and one I want to read in the future.

In this notebook I will use Python to open the file, display what is inside it and find some basic information about the text.

## 1. Reading my text file

First I need to open `my_data.txt` so Python can read the information from it.

```python
# Name of my text file
filename = "my_data.txt"

# Open the file and read what is inside it
with open(filename, "r", encoding="utf-8") as file:
    content = file.read()

print("File loaded successfully!")
```

## 2. Showing what is in the file

Now I will print the file so I can see the list of books.

```python
# Print everything from my text file
print("=== BOOKS I HAVE READ ===")
print()
print(content)
```

## 3. Finding information about my file

Now I will use Python to find some simple information about the file, such as the number of lines, words and characters. I will also find the longest line and show the first and last lines.

```python
# Split the text into separate lines and ignore blank lines
lines = [line.strip() for line in content.splitlines() if line.strip()]

# Split all of the text into separate words
words = content.split()

# Find some information about the file
number_of_lines = len(lines)
number_of_words = len(words)
character_count = len(content)
longest_line = max(lines, key=len)
first_line = lines[0]
last_line = lines[-1]

# Show the results
print("=== FILE STATISTICS ===")
print(f"Number of lines: {number_of_lines}")
print(f"Number of words: {number_of_words}")
print(f"Total character count: {character_count}")
print(f"Longest line: {longest_line}")
print(f"First line: {first_line}")
print(f"Last line: {last_line}")
```

## 4. What I got from the results

The results give me some basic information about my book list. The number of lines shows how many books I included in the file, while the other results show information about the text itself.

Doing this also helped me organise the books in one place and made me recall some books I had read in the past that I might have forgotten about.
