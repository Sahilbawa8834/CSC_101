## Problem 1: Tip Calculator

### Step 1: Understand the problem

The employees at the restaurant need a method for finding the tip to be added to the bill. This system allows users to input the bill value and the tip percentage, and get the values of the tip and the total bill value.

### Step 2: Inputs and outputs

**Input:**
- Bill amount that is a numeric value in dollars (e.g., 45.80$).
- Percentage of tip that is a numeric value (e.g., 15%).

**Output:**
- Amount of tip in dollars up to two decimal places.
- Total bill amount including tip in dollars up to two decimal places.

### Step 3: Work it by hand

Bill: **$45.80**
Tip: **15%**

Step 1: Convert 15% to a decimal
15 ÷ 100 = **0.15**

Step 2: Calculate the tip amount
$45.80 × 0.15 = **$6.87**

Step 3: Total cost
$45.80 + $6.87 = **$52.67**

To the staff member, I will say: **"The tip is $6.87 and the total amount is $52.67."**

## Problem 2: Late Fee Calculator

### Step 1: Understand the problem

The library wants to calculate the fine when someone returns a book late. The fine is 50 cents for each overdue day, but it should never be more than $10.

### Step 2: Inputs and outputs

**Input:**
- Number of days the book is overdue.

**Outputs:**
- Fine amount in dollars, shown to two decimal places.
- If the calculated fine goes above $10, the final fine stays at **$10.00** and the program can show **“Maximum fine reached”**.
- If the book is not overdue (0 days), the fine is **$0.00**.

### Step 3: Work it by hand

**Case A: 5 days overdue**

5 x $0.50 = **$2.50**. Fine is **$2.50**.

**Case B: 25 days overdue**

25 x $0.50 = **$12.50** before the limitation. However, the highest fine that can be levied is $10; hence, the fine is **$10.00**. I will also indicate “Maximum fine exceeded”. This scenario is unique since the calculation exceeds the limitation.

**Case C: 0 days overdue**

0 x $0.50 = **$0.00**. Fine does not apply since the book is on time.
