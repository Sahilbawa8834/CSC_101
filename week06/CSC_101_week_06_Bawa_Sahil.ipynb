# Week 6 - Mini Contact Book

For this worksheet I am making a simple contact book using Python dictionaries. I will create some contacts, add a new one, update a phone number and print all the contacts.

## Part A - Basics

### A1. Create a dictionary with 3 contacts

I will start by making a dictionary where the name is the key and the phone number is the value.

```python
# Create a dictionary with 3 contacts
contacts = {
    "Alice": "0400 123 456",
    "Bob": "0411 222 333",
    "Charlie": "0422 555 777"
}

contacts
```

### A2. Add a new entry

Now I will add myself as another contact.

```python
# Add a new contact
contacts["Sahil"] = "0433 123 456"

contacts
```

### A3. Update an existing contact's number

I can update a value by using the person's name again and giving it a new phone number.

```python
# Change Bob's phone number
contacts["Bob"] = "0411 000 999"

contacts
```

### A4. Print all contacts

I will use a loop with `.items()` to print each name and phone number in a neat format.

```python
# Loop through the dictionary and print each contact
for name, phone in contacts.items():
    print(f"{name}: {phone}")
```

## Part B - Nested Dictionary

For this part each person will have both a phone number and an email. This means each contact will have another dictionary inside it.

```python
# Each contact now has a phone number and an email
contacts = {
    "Alice": {
        "phone": "0400 123 456",
        "email": "alice@example.com"
    },
    "Bob": {
        "phone": "0411 000 999",
        "email": "bob@example.com"
    },
    "Charlie": {
        "phone": "0422 555 777",
        "email": "charlie@example.com"
    },
    "Sahil": {
        "phone": "0433 123 456",
        "email": "sahil@example.com"
    }
}

contacts
```

### B2. Print the contact details

Now I will loop through the nested dictionary and print the name, phone number and email.

```python
# Print the name, phone number and email for each contact
for name, details in contacts.items():
    print(f"Name: {name}")
    print(f"Phone: {details['phone']}")
    print(f"Email: {details['email']}")
    print()
```

### Optional - Look up a contact

I also tried the optional lookup. The user can type a name and the program checks if that person is in the contact book.

```python
# Type a name here to test the lookup
search_name = "Bob"

if search_name in contacts:
    print(f"Name: {search_name}")
    print(f"Phone: {contacts[search_name]['phone']}")
    print(f"Email: {contacts[search_name]['email']}")
else:
    print("Contact not found.")
```
