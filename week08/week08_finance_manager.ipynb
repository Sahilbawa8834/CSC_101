# Week 9 Project: Smart Personal Finance Manager

For this project I made a simple transaction processor. It reads transactions from a text file and checks the data and then shows a summary.

One of the lines in the file is broken on purpose, so the program also needs to skip bad data without crashing the whole program.

## 1. Loading the transaction file

This first function was provided in the project instructions. It opens the file and reads each line. The `try` and `except` also stop the program from crashing if the file cannot be found.

```python
def load_transactions(filename):
    """Read lines from a transaction file safely."""
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()
        return [line.strip() for line in lines if line.strip()]
    except FileNotFoundError:
        print(f"File '{filename}' not found!")
        return []
```

### Testing the file

I will load `transactions.txt` and print the lines first so I can check that the file is being read properly.

```python
# Load the transaction file
filename = "transactions.txt"
raw_lines = load_transactions(filename)

# Show the raw lines from the file
print("Raw transaction data:")
for line in raw_lines:
    print(line)
```

## 2. Processing the transactions

The next function goes through each line and tries to separate the description and amount using the comma.

If the line is good, it is saved as a dictionary. If there is a problem with the line or the amount is not a number, the program skips it instead of stopping.

```python
def process_transactions(lines):
    """Turn transaction lines into dictionaries and skip bad lines."""
    transactions = []
    failed_lines = 0

    for line in lines:
        try:
            # Split the line into description and amount
            description, amount = line.split(",")

            # Remove extra spaces and change the amount into a number
            description = description.strip()
            amount = float(amount.strip())

            # Save the clean transaction as a dictionary
            transaction = {
                "description": description,
                "amount": amount
            }
            transactions.append(transaction)

        except (ValueError, TypeError):
            # Count the line as failed instead of crashing the program
            print(f"Skipped bad line: {line}")
            failed_lines += 1

    return transactions, failed_lines
```

### Testing the processing function

Now I will process the lines. The broken line should be skipped and the other transactions should still work.

```python
# Process the raw data
transactions, failed_lines = process_transactions(raw_lines)

print()
print("Clean transactions:")
for transaction in transactions:
    print(transaction)

print()
print(f"Failed lines: {failed_lines}")
```

## 3. Showing a summary

For the summary I need to count the transactions and calculate the income, expenses and final net amount.

Positive amounts are income and negative amounts are expenses.

```python
def show_summary(transactions, failed_lines):
    """Calculate and display a simple transaction summary."""

    # Add all positive amounts for income
    total_income = sum(
        transaction["amount"]
        for transaction in transactions
        if transaction["amount"] > 0
    )

    # Add all negative amounts for expenses
    total_expenses = sum(
        transaction["amount"]
        for transaction in transactions
        if transaction["amount"] < 0
    )

    # Income plus negative expenses gives the net amount
    net_amount = total_income + total_expenses

    print("=== TRANSACTION SUMMARY ===")
    print(f"Transactions processed: {len(transactions)}")
    print(f"Lines skipped: {failed_lines}")
    print(f"Total income: ${total_income:.2f}")
    print(f"Total expenses: ${abs(total_expenses):.2f}")
    print(f"Net amount: ${net_amount:.2f}")
```

## 4. Putting everything together

The `main()` function connects all the parts together. It loads the file, processes the data and then shows the final results.

```python
def main():
    print("Personal Finance Manager")
    print("=" * 30)

    # Load the data
    filename = "transactions.txt"
    raw_lines = load_transactions(filename)

    if not raw_lines:
        print("No data to process!")
        return

    # Process the data
    transactions, failed_lines = process_transactions(raw_lines)

    # Show the results
    print()
    show_summary(transactions, failed_lines)

# Run the program
main()
```
